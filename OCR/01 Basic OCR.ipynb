Python-tesseract is an optical character recognition (OCR) tool for python. That is, it will recognize and “read” the text embedded in images

```python
from PIL import Image
import pytesseract

# Optional: Set path if Tesseract is not in your PATH (Windows example)
# pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# Load image

image = Image.open(r"D:\mock_thesis\temp\sample_scan.png")

# Use Tesseract to extract text
text = pytesseract.image_to_string(image)

# print("....................................")
# print(text)


# Convert to grayscale and apply thresholding to clean up the image
gray_image = image.convert("L")
threshold = gray_image.point(lambda x: 0 if x < 140 else 255, '1')

text = pytesseract.image_to_string(threshold)
print("....................................")
print(text)

```
